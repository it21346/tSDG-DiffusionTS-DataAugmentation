```python
import pandas as pd
import numpy as np
ori_data = pd.read_csv('ETTh1.csv', sep=',')

ori_data.rename(columns={'date': 'Timestamp'}, inplace=True)
ori_data['Timestamp'] = pd.to_datetime(ori_data["Timestamp"])


h = ori_data['Timestamp'].dt.hour
ori_data['hour_sin'] = np.sin(2*np.pi * h / 24)
ori_data['hour_cos'] = np.cos(2*np.pi * h / 24)    
ori_data['month_sin'] = np.sin(2 * np.pi * ori_data['Timestamp'].dt.month / 12)
ori_data['month_cos'] = np.cos(2 * np.pi * ori_data['Timestamp'].dt.month / 12)
ori_data['day_of_week_sin'] = np.sin(2 * np.pi * ori_data['Timestamp'].dt.dayofweek / 7)
ori_data['day_of_week_cos'] = np.cos(2 * np.pi * ori_data['Timestamp'].dt.dayofweek / 7)


ori_data.drop(columns=['Timestamp'], inplace=True)

column_to_move = 'OT'
ori_data = ori_data[[col for col in ori_data.columns if col != column_to_move] + [column_to_move]]
ori_data.to_csv('ETTh1_energy_data.csv', index=False)
```
